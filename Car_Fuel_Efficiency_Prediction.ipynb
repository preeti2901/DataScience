```python
Predict car mileage (MPG) based on engine size
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Simplified clean dataset
data = {
    "Engine_Size": [130,165,150,140,198,220,215,225,190,170],
    "MPG": [18,15,18,16,15,14,14,14,15,16]
}

df = pd.DataFrame(data)
df
```

```python
#Define X and y

X = df[["Engine_Size"]]
y = df["MPG"]
```

```python
#Train model

model = LinearRegression()
model.fit(X,y)
```

```python
#Value Prediction

predict_val = model.predict([[250]])
print("Predicted value: ",predict_val)
```

```python
#Visual

plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Engine_Size")
plt.ylabel("MPG")
plt.title("Engine_Size vs Size")
plt.show()
```
